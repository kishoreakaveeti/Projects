AZ-500 is an **implementation security exam**, not a pure architecture exam. As a solution architect, you may know the right service at a high level, but the exam often tests exact knobs:

- Which role gives the minimum required permission?
- Which network control applies to inbound traffic vs outbound traffic?
- Which encryption feature protects data at rest vs data in use vs data in transit?
- Which logging/data collection mechanism is needed before Sentinel can alert?
- Which Defender plan enables which feature?
- Which identity type avoids storing secrets?
- Which control prevents deletion/modification vs only detects it?
- Which approach supports private access without public endpoint exposure?

## Most common answer-selection principles

When two answers look correct, prefer the answer that:

1. Uses **managed identity** instead of client secret or connection string.
2. Uses **RBAC least privilege** instead of Owner/Contributor.
3. Uses **Private Endpoint / Private Link** instead of public endpoints where private access is required.
4. Uses **Conditional Access** for user sign-in enforcement, not RBAC.
5. Uses **PIM** for just-in-time privileged role activation, not permanent admin assignment.
6. Uses **Azure Policy / initiative** for governance enforcement at scale.
7. Uses **Defender for Cloud** for posture and workload protection.
8. Uses **Microsoft Sentinel** for SIEM/SOAR detection, analytics, incidents, automation.
9. Uses **Key Vault** for secrets/keys/certificates, preferably with RBAC and private access.
10. Uses **WAF** for HTTP/S application-layer attacks, not NSG.

# 2. Secure identity and access — deep dive

This domain is only 15–20%, but it is foundational. Many wrong AZ-500 answers come from mixing up four different authorization systems:

1. **Microsoft Entra ID roles** — control administration of identity objects such as users, groups, applications, Conditional Access, and directory settings.
2. **Azure RBAC roles** — control management-plane and data-plane access to Azure resources.
3. **Application permissions / OAuth grants** — control what applications can do against APIs such as Microsoft Graph.
4. **Resource-local authorization** — examples: SQL permissions, Key Vault access model, Storage ACLs, Kubernetes RBAC.

## 2.1 Identity terms you must separate

| Term | Meaning | Exam trap |
|---|---|---|
| Microsoft Entra tenant | Identity boundary/directory | Not the same as Azure subscription. |
| Azure subscription | Billing/resource boundary | Uses Entra tenant for identity but has its own RBAC. |
| User | Human identity | MFA/Conditional Access applies to users/sign-ins. |
| Group | Collection of identities | Prefer groups for role assignments at scale. |
| Service principal | Application identity instance in tenant | Often confused with app registration. |
| App registration | Application definition/configuration | Defines redirect URIs, certificates/secrets, permissions. |
| Enterprise application | Service principal representation | Used for app access, assignment, SSO, consent. |
| Managed identity | Azure-managed service principal | Best answer when app runs on Azure and needs resource access. |
| Workload identity federation | Allows workload to use external identity without secrets | Relevant for GitHub/Azure DevOps/OIDC scenarios. |

## 2.2 Azure RBAC model

Azure RBAC assignment = **security principal + role definition + scope**.

### Security principal examples
- User
- Group
- Service principal
- Managed identity

### Scope levels
- Management group
- Subscription
- Resource group
- Resource

Access inherited downward. A role assigned at subscription applies to resource groups and resources unless denied through another mechanism such as deny assignments/managed apps/Azure Blueprints legacy patterns.

### Built-in roles you must know

| Role | What it can do | What it cannot do / trap |
|---|---|---|
| Owner | Full resource management and can assign roles | Too broad for least privilege. |
| Contributor | Full resource management | Cannot assign Azure RBAC roles. |
| Reader | View resources | Cannot view secrets by default. |
| User Access Administrator | Manage Azure RBAC assignments | Can grant access; high privilege. |
| Security Admin | View and update security policies; manage security components | Not general Owner. |
| Security Reader | View security state | Read-only. |
| Key Vault Administrator | Manage Key Vault and objects under RBAC model | Does not grant broad subscription rights. |
| Key Vault Secrets User | Read secret contents | Data-plane permission. |
| Storage Blob Data Contributor | Read/write/delete blob data | Different from Storage Account Contributor. |
| Storage Account Contributor | Manage storage account | Does not automatically read blob data when Entra data-plane is required. |
| Network Contributor | Manage network resources | Does not manage VMs/storage. |
| Virtual Machine Contributor | Manage VMs | Does not grant access to virtual network or storage account. |
| AcrPull | Pull images from ACR | Use for AKS/kubelet pull access. |
| AcrPush | Push/pull images | For build pipelines, not runtime if pull-only required. |

## 2.3 Azure RBAC vs data-plane RBAC

A major trap: management-plane access is not the same as data access.

Example:
- **Storage Account Contributor** can manage storage account settings, but for blob data access using Entra ID you need roles like **Storage Blob Data Reader/Contributor/Owner**.
- **Key Vault Contributor** can manage Key Vault resource settings, but cannot necessarily read secrets/keys/certificates unless data-plane permissions are granted.
- **SQL Server Contributor** can manage Azure SQL server settings, but database-level access still depends on SQL/Entra database authentication and permissions.

## 2.4 Custom roles

Use custom roles when no built-in role matches least privilege.

Custom role components:
- `Name`
- `Description`
- `Actions` — allowed management-plane operations.
- `NotActions` — excluded management-plane operations.
- `DataActions` — allowed data-plane operations.
- `NotDataActions` — excluded data-plane operations.
- `AssignableScopes` — where role can be assigned.

### Trap
`NotActions` is not a deny. If a user has another role that grants the action, they may still perform it.

## 2.5 Privileged Identity Management — PIM

PIM is used for just-in-time privileged access. It supports eligible assignments, activation, approval, justification, MFA, notifications, access reviews, and auditing.

### Key concepts

| Concept | Meaning |
|---|---|
| Eligible assignment | User can activate role when needed. |
| Active assignment | User currently has the role. |
| Permanent eligible | Can activate indefinitely over time. |
| Time-bound eligible | Eligibility expires. |
| Activation duration | How long role stays active after activation. |
| Approval | Another approver must approve activation. |
| Justification | User must enter business reason. |
| MFA on activation | User must satisfy MFA before role activation. |

### When PIM is the answer
- Admin rights must be temporary.
- Privileged access requires approval.
- Role activation should require justification/MFA.
- Need audit history of privileged activations.
- Reduce standing privileges.

### When PIM is not the answer
- You need to block sign-in from risky countries — use Conditional Access.
- You need to grant a workload access to Storage — use managed identity + RBAC.
- You need to enforce resource configurations — use Azure Policy.

## 2.6 Conditional Access

Conditional Access controls whether a sign-in is allowed and what controls are required.

Policy structure:
1. **Assignments**: users/groups/workload identities, cloud apps/actions.
2. **Conditions**: risk, device platform, location, client apps, device filters.
3. **Access controls**: grant/block, require MFA, require compliant device, require hybrid joined device, require approved app, require app protection policy.
4. **Session controls**: sign-in frequency, persistent browser session, app-enforced restrictions, Conditional Access App Control.

### Common CA scenarios

| Requirement | Likely answer |
|---|---|
| Require MFA for Azure portal | Conditional Access targeting Microsoft Azure Management. |
| Block legacy authentication | Conditional Access client apps condition or authentication policies. |
| Require compliant device for admin portal | Conditional Access with require compliant device. |
| Block access from named countries | Conditional Access named locations. |
| Require stronger auth for risky sign-ins | Conditional Access + Identity Protection risk. |
| Restrict session in SaaS app | Conditional Access session controls / Defender for Cloud Apps. |

### Trap
RBAC controls what you can do **after authentication**. Conditional Access controls whether/how sign-in is allowed. Do not use RBAC to enforce MFA.

## 2.7 MFA

MFA can be enforced through:
- Conditional Access policy — modern/preferred.
- Security defaults — simple baseline, less granular.
- Per-user MFA — legacy, avoid for new designs unless specifically asked.
- PIM activation settings — require MFA before role activation.

## 2.8 App registrations, enterprise apps, OAuth consent

### App registration
Defines an application in Entra ID:
- Application/client ID.
- Redirect URI.
- Certificates and secrets.
- API permissions.
- Exposed API scopes.
- App roles.
- Authentication settings.

### Enterprise application
The service principal instance in a tenant:
- User/group assignment.
- SSO configuration.
- Conditional Access targeting.
- Permissions consent records.

### Delegated vs application permissions

| Permission type | Used when | Identity used | Example |
|---|---|---|---|
| Delegated | App acts on behalf of signed-in user | User + app | App reads user's profile. |
| Application | App acts as itself without user | App/service principal | Daemon reads all users. |

### Admin consent vs user consent
- User consent: user consents to low-risk delegated permissions if allowed.
- Admin consent: tenant admin approves permissions, especially application permissions or high privilege scopes.

## 2.9 Managed identities

Managed identity is a Microsoft-managed service principal for Azure resources.

| Type | Lifecycle | Best use |
|---|---|---|
| System-assigned | Tied to one Azure resource; deleted with resource | Single app/VM/function needing access. |
| User-assigned | Independent Azure resource; can attach to many resources | Shared identity across multiple resources or lifecycle separation. |

### Why managed identity is often the best exam answer
- No secrets in code.
- No manual credential rotation.
- Works with Azure RBAC and many Azure services.
- Supports least privilege through scoped role assignments.

### Trap
Managed identity only works for resources/services that support it. For external apps, use service principal, certificate credential, or workload identity federation.


````python
# Azure CLI snippets — Identity and RBAC
# These commands are reference snippets. Run only after setting your variables and logging in.

# Login and select subscription
az login
az account set --subscription "<subscription-id-or-name>"

# Show current account
az account show --output table

# List role definitions containing Storage Blob
az role definition list --name "Storage Blob Data Contributor" --output jsonc

# Assign a built-in role to a user/group/service principal/managed identity
az role assignment create   --assignee "<object-id-or-upn-or-appid>"   --role "Storage Blob Data Contributor"   --scope "/subscriptions/<sub-id>/resourceGroups/<rg-name>/providers/Microsoft.Storage/storageAccounts/<storage-account>"

# List role assignments for a principal
az role assignment list --assignee "<object-id>" --all --output table

# Create a custom role from JSON
az role definition create --role-definition ./custom-role.json

# Update a custom role
az role definition update --role-definition ./custom-role.json

# Delete a custom role
az role definition delete --name "<custom-role-name>"
````

## 2.10 Identity exam traps

### Trap 1 — Azure RBAC vs Microsoft Entra roles

**Scenario:** User must create Conditional Access policies.  
**Wrong thinking:** Assign Contributor on subscription.  
**Correct thinking:** Conditional Access is an Entra ID function. Use appropriate Entra role such as Conditional Access Administrator or Security Administrator, depending on requirement.

### Trap 2 — Contributor cannot assign roles

**Scenario:** User can manage all resources and must grant access to others.  
**Correct answer:** Owner or User Access Administrator. Contributor is not enough.

### Trap 3 — Storage Account Contributor cannot necessarily read blob data

For blob data operations with Entra authorization, use Storage Blob Data Reader/Contributor/Owner.

### Trap 4 — PIM vs Conditional Access

- PIM: temporary privileged role activation.
- Conditional Access: sign-in requirements such as MFA, device compliance, location, risk.

### Trap 5 — App registration vs enterprise application

- App registration = app definition.
- Enterprise application = service principal instance used for assignment, SSO, consent, Conditional Access targeting.

### Trap 6 — Client secret vs certificate vs managed identity

Preferred order for Azure-hosted workload:
1. Managed identity.
2. Workload identity federation.
3. Certificate credential.
4. Client secret only when unavoidable.

### Trap 7 — `NotActions` is not deny

Azure RBAC is additive. `NotActions` excludes actions from that specific role definition but does not block another role from granting them.

## Identity mini case study

**Requirement:** A developer team needs to deploy App Services to a resource group. They must not assign RBAC permissions. The app must read secrets from Key Vault without storing credentials.

**Solution pattern:**
- Assign team **Contributor** at resource group scope, not Owner.
- Enable **managed identity** on App Service.
- Grant managed identity **Key Vault Secrets User** on Key Vault or selected scope.
- Store app secrets in Key Vault.
- Avoid client secrets in app settings.


# 3. Secure networking — deep dive

Networking is a high-value AZ-500 domain because the exam gives scenario-based questions with overlapping products. You need crisp service boundaries.

## 3.1 Network security service decision table

| Requirement | Best-fit service/control | Why |
|---|---|---|
| Allow/deny traffic at subnet/NIC level by IP/port/protocol | NSG | Stateful L3/L4 filtering. |
| Simplify NSG rules by grouping VMs | ASG | Logical app groups for NSG rules. |
| Force traffic through firewall/NVA | UDR / route table | Controls next hop. |
| Central managed firewall with FQDN/application/network rules | Azure Firewall | Stateful cloud firewall with threat intelligence and policies. |
| Protect HTTP/S app from OWASP attacks | WAF | Application-layer web protection. |
| Regional layer 7 load balancer with WAF | Application Gateway WAF | Regional ingress. |
| Global HTTP/S acceleration and WAF | Azure Front Door Premium/Standard + WAF | Global edge entry point. |
| Private access to PaaS through private IP | Private Endpoint | Removes need for public endpoint. |
| Restrict PaaS public endpoint to selected VNet subnet | Service Endpoint | Still uses public endpoint but source is VNet identity. |
| Expose your own service privately to consumers | Private Link Service | Provider-side private service publishing. |
| Hybrid S2S encrypted tunnel over internet | VPN Gateway S2S | IPsec/IKE tunnel. |
| User remote VPN to Azure | VPN Gateway P2S | Individual client VPN. |
| Private dedicated circuit to Microsoft | ExpressRoute | Private connectivity, not encrypted by default. |
| Central branch/cloud connectivity | Virtual WAN | Hub-and-spoke at scale. |
| Central security policy for hubs/firewalls | Firewall Manager | Manage Firewall policies, secured virtual hubs. |
| DDoS volumetric protection for public IPs | Azure DDoS Protection | Enhanced L3/L4 DDoS protection. |
| Diagnose connectivity | Network Watcher | IP flow verify, next hop, NSG flow logs, connection troubleshoot. |

## 3.2 NSG fundamentals

Network Security Groups filter inbound/outbound traffic at subnet and/or NIC.

### NSG rule properties
- Priority: 100–4096, lower number wins.
- Direction: Inbound or Outbound.
- Source: IP, CIDR, service tag, ASG.
- Source port range.
- Destination: IP, CIDR, service tag, ASG.
- Destination port range.
- Protocol: TCP, UDP, ICMP, ESP, AH, Any.
- Action: Allow or Deny.

### Default NSG rules
Inbound:
- AllowVNetInBound.
- AllowAzureLoadBalancerInBound.
- DenyAllInBound.

Outbound:
- AllowVNetOutBound.
- AllowInternetOutBound.
- DenyAllOutBound.

### Exam traps
- NSGs are stateful. If inbound allowed, response outbound is allowed automatically.
- Lower priority number wins.
- NSG can be associated with subnet and NIC. Both must allow traffic.
- ASG only applies to NIC-backed resources, not arbitrary PaaS resources.
- NSG is not WAF and cannot inspect HTTP payloads for SQL injection/XSS.

## 3.3 ASG — Application Security Groups

ASGs let you group VM NICs by application role. Instead of writing NSG rules using IP addresses, you use source/destination ASGs.

Example:
- `asg-web`
- `asg-app`
- `asg-db`

Rule: allow TCP 443 from `asg-web` to `asg-app`; allow TCP 1433 from `asg-app` to `asg-db`.

## 3.4 UDR and route tables

User-defined routes override Azure default routing for subnets. Common use:
- Force all outbound traffic `0.0.0.0/0` to Azure Firewall/NVA.
- Route spoke-to-spoke traffic through hub firewall.
- Route on-premises prefixes to virtual network gateway.

### Next hop types
- Virtual appliance
- Virtual network gateway
- VNet peering
- Internet
- None

### Trap
UDRs influence routing. They do not filter traffic. Use NSG/firewall to allow/deny.

## 3.5 Azure Firewall

Azure Firewall is a managed stateful firewall. It supports:
- Network rules: L3/L4, IP/port/protocol.
- Application rules: FQDN-based outbound HTTP/S/MSSQL filtering.
- DNAT rules: inbound NAT to private resources.
- Threat intelligence filtering.
- TLS inspection in Premium.
- IDPS in Premium.
- Web categories.
- Firewall Policy.

### Azure Firewall vs NSG

| Feature | NSG | Azure Firewall |
|---|---|---|
| Scope | Subnet/NIC | Central firewall resource |
| L3/L4 filtering | Yes | Yes |
| FQDN outbound filtering | Limited via service tags only | Yes application rules |
| Threat intelligence | No | Yes |
| DNAT | No | Yes |
| Central logging/policy | Limited | Strong |
| TLS inspection | No | Premium |

## 3.6 Application Gateway, WAF, Front Door

### Application Gateway
Regional layer 7 load balancer. Supports:
- Path-based routing.
- Multi-site hosting.
- TLS termination/end-to-end TLS.
- Cookie-based affinity.
- WAF_v2.
- Private frontend IP.

### Azure Front Door
Global HTTP/S entry point. Supports:
- Global anycast edge.
- Acceleration.
- WAF.
- Rules engine.
- Origin groups/failover.
- Private Link to origins in Premium.

### WAF
Web Application Firewall protects against web attacks such as:
- SQL injection.
- Cross-site scripting.
- OWASP Core Rule Set threats.
- Bot protection, custom rules, managed rules.

### Exam decision
- Regional app behind Azure VNet: Application Gateway WAF.
- Global public app with edge acceleration/failover: Front Door WAF.
- Network-wide outbound filtering: Azure Firewall.
- Subnet/NIC filtering: NSG.

## 3.7 Private Endpoint vs Service Endpoint vs Private Link Service

This is one of the most important AZ-500 topics.

| Feature | Private Endpoint | Service Endpoint | Private Link Service |
|---|---|---|---|
| Purpose | Access Azure PaaS privately through private IP in your VNet | Secure public PaaS endpoint to selected VNet subnet | Publish your own service privately to consumers |
| Public endpoint required? | Can disable public network access | Yes, service still has public endpoint | Provider service exposed privately |
| DNS complexity | Requires private DNS integration | Less DNS complexity | Provider/consumer setup |
| Data path | Microsoft backbone using private IP | Microsoft backbone to public service endpoint | Private Link |
| Best for | Storage, Key Vault, SQL, ACR private access | Simpler subnet-based restriction | SaaS/provider service privately exposed |

### Private Endpoint traps
- Private endpoint creates a NIC in your VNet.
- DNS must resolve service FQDN to private IP, usually using private DNS zones.
- NSGs historically had restrictions but now can be enabled for private endpoints depending on configuration; exam usually focuses on private access concept.
- Private Endpoint is preferred when public access must be disabled.

### Service Endpoint traps
- Service endpoint does not give the PaaS resource a private IP in your VNet.
- It restricts access to selected virtual networks/subnets but the service endpoint remains public.
- For “no public exposure,” choose Private Endpoint.

## 3.8 App Service networking

App Service has several networking features:

| Requirement | Feature |
|---|---|
| App needs outbound access to VNet resources | VNet Integration |
| App should be reachable privately from VNet | Private Endpoint |
| Lock down inbound to selected IPs/VNets | Access restrictions / private endpoint |
| Fully isolated dedicated environment | App Service Environment v3 |
| Force all outbound through VNet/firewall | Regional VNet Integration + route all |

### Trap
VNet Integration is mostly for outbound traffic from App Service into your VNet. It does not by itself make the app privately reachable inbound. Use Private Endpoint for private inbound access.

## 3.9 VPN, ExpressRoute, Virtual WAN

### VPN Gateway

## 3.12 Networking exam traps

### Trap 1 — Private Endpoint vs Service Endpoint

Requirement says “must not be accessible over the public internet” or “private IP in VNet” → **Private Endpoint**.

Requirement says “allow only selected subnet to access service endpoint” and public endpoint can remain → **Service Endpoint** may fit.

### Trap 2 — WAF vs Azure Firewall

- WAF: HTTP/S application-layer protection.
- Azure Firewall: network/application firewall for broader traffic control.

### Trap 3 — Application Gateway vs Front Door

- Application Gateway: regional, VNet-based layer 7.
- Front Door: global edge entry point.

### Trap 4 — VNet Integration vs Private Endpoint for App Service

- VNet Integration: outbound from App Service to VNet.
- Private Endpoint: inbound private access to App Service.

### Trap 5 — ExpressRoute encryption

ExpressRoute is private connectivity but not automatically encrypted end-to-end. If requirement says encryption over ExpressRoute, think IPsec VPN over ExpressRoute/private peering or MACsec depending context.

### Trap 6 — NSG does not inspect web payloads

If the requirement mentions OWASP, SQL injection, cross-site scripting, bot, HTTP header/path rules → WAF, not NSG.

### Trap 7 — UDR does not filter

UDR controls route path; firewall/NSG controls allow/deny.

| Feature | Protects against |
|---|---|

| Feature | Meaning |
|---|---|

# 5.1 Azure Policy

Azure Policy evaluates resources for compliance against rules. Policies can audit, deny, modify, deploy supporting resources, or mark as disabled.

## Key terms

| Term | Meaning |
|---|---|

## 5.11 Governance/Defender/Sentinel exam traps

### Trap 1 — Policy vs Defender

- Azure Policy enforces/evaluates configurations.
- Defender for Cloud surfaces security recommendations and secure score; many recommendations are backed by policy.

### Trap 2 — Defender vs Sentinel

- Defender for Cloud: posture + workload protection for cloud resources.
- Sentinel: SIEM/SOAR for collecting, correlating, detecting, investigating across many sources.

### Trap 3 — Diagnostic setting vs DCR

- Diagnostic settings route Azure resource logs/metrics.
- DCRs configure collection for Azure Monitor Agent and some pipelines.

### Trap 4 — Log Analytics workspace is the data store

Sentinel is enabled on a Log Analytics workspace. If logs are not connected/ingested, Sentinel analytics cannot detect from them.

### Trap 5 — Automation rule vs playbook

- Automation rule: Sentinel-native incident automation conditions/actions.
- Playbook: Logic App workflow.

### Trap 6 — Audit vs Deny policy

- Audit reports non-compliance.
- Deny prevents non-compliant deployment/update.

### Trap 7 — DeployIfNotExists requires remediation for existing resources

A DeployIfNotExists policy can deploy related resources for new/evaluated resources, but existing non-compliant resources typically need a remediation task.

````python
// Azure Policy skeleton — deny resources not in allowed locations
{
  "mode": "All",
  "parameters": {
    "allowedLocations": {
      "type": "Array",
      "metadata": {
        "displayName": "Allowed locations",
        "description": "The list of allowed locations for resources."
      }
    }
  },
  "policyRule": {
    "if": {
      "not": {
        "field": "location",
        "in": "[parameters('allowedLocations')]"
      }
    },
    "then": {
      "effect": "deny"
    }
  }
}
````
