# 2. Python, JSON, HTTP, and cloud from zero

A **program** executes instructions. A **function** packages an operation. A **dictionary** maps names to values. A **list** preserves an ordered collection. JSON is a portable data representation; Python `None`, `True`, and `False` become JSON `null`, `true`, and `false`. JSON does not execute a function: a tool call is data until an application deliberately dispatches it.

An **API** is an agreed interface. HTTP carries a method, URL, headers, and often a body. `POST /v1/messages` asks a service to create a model response. A response includes a status and body. A 200 response tells you transport succeeded, not that the answer is factually correct. A 401 often means credentials are missing or invalid; a 403 means permissions are insufficient; a 429 signals throttling; a 5xx signals a server-side problem. Exact API categories are provider-specific.

**Authentication** asks who is calling; **authorization** asks what that identity may do. A credential proving you are Kishore does not entitle you to read every account. A tool input saying `tenant_id="B"` must never override the authenticated caller's tenant.

Cloud computing supplies infrastructure through managed services. Compute runs your application; storage preserves files; databases preserve queryable state; a network connects them. Managed serverless compute reduces server administration but still has timeouts, concurrency limits, permission configuration, and cost. A container packages code and dependencies; orchestration schedules container instances. “Serverless” does not mean unlimited execution or no servers.

Keep the model and your application conceptually separate. A typical system has a client UI, application API, authentication layer, orchestrator, retrieval store, model endpoint, tool services, audit trail, and monitoring. The model service need not be in the same account or region as your database; crossing those boundaries creates data-flow and latency decisions.

**State** is information that must survive calls: conversation history, job status, approvals, and completed transactions. A Python variable survives only while its process survives. A production approval should be persisted with identity, scope, expiry, and exact action parameters.

# 3. What an LLM does—and what an agent adds

An LLM receives a sequence of tokens and predicts a continuation. A token is a unit of the model's text encoding, not necessarily a word or character. A message is encoded with role and formatting information. A context window bounds the information usable in one model invocation; persistent memory is a separate application design.

Let $x_{1:t}$ be the token prefix. The model represents a distribution $P(x_{t+1}\mid x_{1:t})$. If candidate logits are $z_i$, softmax produces

$$P_i=\frac{\exp(z_i/T)}{\sum_j\exp(z_j/T)}.$$

Here $T>0$ is temperature in this conceptual example. Lower temperature concentrates the distribution; higher temperature spreads it. Do not infer that every current Claude model exposes every sampling setting, or that low temperature makes a false premise true. Choosing a higher-probability continuation is not equivalent to verifying a database record.

Training modifies model parameters using datasets and optimization. Inference uses those parameters to produce an output. Tool definitions are normally supplied **at inference time**; your application does not have to retrain Claude each time you add an order API. The model has learned patterns of structured interaction, then conditions on your tool names, descriptions, schemas, and conversation. Your code determines whether the requested operation actually runs.

An **augmented LLM** combines a model with retrieval or tools. A **workflow** has application-defined steps. An **agent** allows the model to choose some next steps within an application harness. A harness provides tools, memory, permissions, budgets, error handling, and execution. “An agent is an LLM” omits those parts. “All tool use is autonomous agency” is also too broad: a fixed extraction call can use a tool schema without an open-ended loop.

Use the smallest architecture that meets the requirement. A single prompt may classify a ticket. A workflow may extract fields then run arithmetic validation. An agent may investigate an uncertain billing issue. Adding agents increases coordination burden, token consumption, and failure paths. It is justified by task structure, not by novelty.

**Worked interpretation:** An answer has 95% probability under the model. That does not establish 95% factual correctness; probability is conditioned on the supplied text and learned distribution. For decision confidence, calibrate against held-out outcomes and evidence.

**Pitfalls:** confusing token limit with storage; treating model text as authenticated state; assuming tool access is training; treating confident language as measured reliability.

Further reading: [agent design patterns](https://www.anthropic.com/engineering/building-effective-agents), [models overview](https://platform.claude.com/docs/en/about-claude/models/overview).


````python
def softmax(logits, temperature=1.0):
    if temperature <= 0:
        raise ValueError("temperature must be positive")
    shifted = [(z-max(logits))/temperature for z in logits]
    weights = [math.exp(z) for z in shifted]
    return [w/sum(weights) for w in weights]

for temp in [0.3, 1.0, 2.0]:
    print(temp, [round(p, 4) for p in softmax([3, 2, 0], temp)])
````

# 4. Messages API: the complete request and response

The Messages API is the foundation of a direct Claude integration. Your request identifies a model, an output limit, and messages. The system instruction is a top-level parameter, not a fabricated `system` message inside the user/assistant list. `max_tokens` controls an output ceiling; it does not tell the API to forget long input.

The response contains content blocks, a stop reason, and usage. Blocks may contain text, tool requests, or other supported types. Never assume `response.content[0]` is always plain text. Retain full assistant blocks when continuing a tool exchange, rather than extracting only the readable sentence.

| Stop signal | Meaning for an application |
|---|---|
| `end_turn` | The model has finished this response turn |
| `tool_use` | Dispatch authorized client tools, return their results, continue |
| `max_tokens` | Output was cut at its configured limit; treat completeness cautiously |
| `stop_sequence` | A configured stop sequence ended generation |
| Other supported signals | Consult the current stop-reason docs; do not silently treat as success |

Finishing a response is not proof that a business transaction succeeded. Your business workflow can remain incomplete after `end_turn`. A text statement such as “done” is not a protocol termination condition.

A direct Messages integration must provide the relevant conversation or a carefully reconstructed context on subsequent calls. Previous responses do not automatically reappear in a new request. This differs from higher-level session abstractions that maintain conversation state for you.

The optional code below uses standard-library HTTP so you can see the actual wire shape. It sets explicit timeouts, does not print the API key, and does not call anything unless you enable `RUN_LIVE`. For practical projects, install the official `anthropic` SDK in a virtual environment and use its response types, timeout and retry configuration, and request IDs.

# 5. Tool contracts and schema design

A tool is an operation the model may request. The interface must state what it does, when it is appropriate, its required input, important boundaries, and the nature of its output. “Handles orders” is too vague for choosing between searching orders, fetching one order, and refunding one order.

Prefer a focused contract such as `get_order(order_id)` over `do_anything(query)`. Focused contracts are easier to validate, authorize, test, and observe. Do not create hundreds of redundant tools; scope discovery to the task and role. A synthesizer usually needs evidence, not unrestricted shell access.

JSON Schema declares structure. `properties` describes fields; `required` requires presence; `additionalProperties: false` rejects undeclared fields where supported; `enum` limits values. A required field can still be nullable if its type permits `null`. A missing value and a zero value differ. For extraction, `null` can honestly represent an absent invoice date; an invented date cannot.

Schema correctness, factual correctness, and permission correctness are separate. Strict tool use can constrain inputs on supported models and schemas. It cannot prove that an order belongs to the caller or that an invoice total is mathematically consistent. Application code enforces those rules.

Tool choice can allow free selection (`auto`), require some tool (`any`), or require a named tool (`tool`). Forcing a named extraction is useful when the task is known. Forcing a call is **not** equivalent to authorizing an irreversible action, and it does not replace prerequisite gates.

Error contracts should distinguish an empty successful search from failed access. A structured error can include category, retryability, stable code, a safe explanation, and correlation ID. The model should not receive stack traces containing secrets. A permission failure usually needs a scope fix or escalation, not five identical retries.

**Worked contract:** `get_order` accepts one opaque ID and returns only return-relevant fields. The caller identity is supplied by trusted application context, never an editable model argument. `prepare_refund` returns a proposal. `commit_refund` is a separately authorized state-changing operation.

References: [define tools](https://platform.claude.com/docs/en/agents-and-tools/tool-use/define-tools), [strict tool use](https://platform.claude.com/docs/en/agents-and-tools/tool-use/strict-tool-use).


````python
ORDER_TOOL = {
    "name": "get_order",
    "description": "Read return-relevant details for one known order ID. Does not search customers or issue refunds.",
    "input_schema": {
        "type": "object", "properties": {"order_id": {"type": "string"}},
        "required": ["order_id"], "additionalProperties": False}}

# A tiny illustrative validator; this is NOT a general JSON Schema implementation.
def validate_order_args(args):
    if not isinstance(args, dict) or set(args) != {"order_id"}:
        raise ValueError("Exactly order_id is required")
    if not isinstance(args["order_id"], str) or not args["order_id"].startswith("O-"):
        raise ValueError("order_id must be a known-format opaque string")

validate_order_args({"order_id": "O-101"})
try:
    validate_order_args({"order_id": "O-101", "tenant_id": "other"})
except ValueError as e:
    print("Rejected:", e)
print(json.dumps(ORDER_TOOL, indent=2))
````

# 6. A runnable agent loop, step by step

The client-tool round trip has five stages: send messages and tool schemas; receive assistant blocks; inspect the stop reason; validate and execute each requested tool; append one result for every tool-use ID, then call the model again. The result block associates evidence with the exact request by ID.

For a tool-use response, preserve the entire assistant content, including any surrounding text. Then add a user message containing the `tool_result` blocks. The model receives observations and decides what to do next. If several tool calls arrive together, return results for each and preserve dependencies. Independent reads can run concurrently; a dependent write must wait for its prerequisites.

Stopping has two layers. Protocol completion uses `end_turn`; resource protection uses a maximum number of turns, deadline, cost budget, or cancellation. A resource cap produces **incomplete** status and escalation, not a claim that the job completed. A model may repeatedly request a tool because a result is malformed; the cap protects you while telemetry helps diagnose the cause.

The following fake model deliberately emits a deterministic sequence. It is a protocol simulator, not a trained LLM. It lets you inspect the mechanics without API cost. The executor validates names and arguments and records failures explicitly. Use this loop as a learning starting point; production needs persistence, concurrency controls, authorization, request tracing, and more complete stop-reason handling.

**Pitfalls:** parsing “I am finished”; stopping because one text block exists alongside a tool block; forgetting assistant blocks; mismatching IDs; exposing every registered function to every agent; silently treating unknown tools as success.

**Exercise:** Modify the fake model to request an unknown tool. Confirm that it receives an error observation rather than running arbitrary Python. Then modify it to never stop: observe the incomplete result from the budget guard.

Reference: [tool protocol overview](https://platform.claude.com/docs/en/agents-and-tools/tool-use/overview), [handling tool calls](https://platform.claude.com/docs/en/agents-and-tools/tool-use/handle-tool-calls).


````python
def fake_model(messages, tools):
    if not any(any(b.get("type") == "tool_result" for b in m["content"])
               for m in messages if isinstance(m.get("content"), list)):
        return {"stop_reason": "tool_use", "content": [
            {"type": "text", "text": "I will inspect the order."},
            {"type": "tool_use", "id": "toolu_lab_1", "name": "get_order",
             "input": {"order_id": "O-101"}}]}
    return {"stop_reason": "end_turn", "content": [
        {"type": "text", "text": "Order O-101 is delivered. No refund was executed."}]}

def run_agent(model_call, messages, tool_definitions, handlers, max_turns=4):
    history = json.loads(json.dumps(messages))
    exposed = {t["name"] for t in tool_definitions}
    trace = []
    for turn in range(max_turns):
        response = model_call(history, tool_definitions)
        history.append({"role": "assistant", "content": response["content"]})
        trace.append({"turn": turn, "stop_reason": response["stop_reason"]})
        if response["stop_reason"] == "end_turn":
            return {"status": "complete", "history": history, "trace": trace}
        if response["stop_reason"] != "tool_use":
            return {"status": "incomplete", "reason": response["stop_reason"], "trace": trace}
        calls = [b for b in response["content"] if b["type"] == "tool_use"]
        if not calls:
            return {"status": "incomplete", "reason": "malformed_tool_turn", "trace": trace}
        results = []
        for call in calls:
            try:
                if call["name"] not in exposed or call["name"] not in handlers:
                    raise ValueError("Tool not allowed")
                validate_order_args(call["input"])
                result = handlers[call["name"]](**call["input"])
                results.append({"type": "tool_result", "tool_use_id": call["id"],
                                "content": json.dumps(result)})
            except (ValueError, PermissionError) as error:
                results.append({"type": "tool_result", "tool_use_id": call["id"],
                                "is_error": True, "content": json.dumps({
                                    "category": "validation_or_permission", "retryable": False,
                                    "message": str(error)})})
        history.append({"role": "user", "content": results})
    return {"status": "incomplete", "reason": "turn_budget", "history": history, "trace": trace}

agent_result = run_agent(fake_model,
    [{"role": "user", "content": "Inspect order O-101"}],
    [ORDER_TOOL], {"get_order": local_order_api})
assert agent_result["status"] == "complete"
print(json.dumps(agent_result, indent=2))
````

# 7. Enforcement, authorization, and irreversible actions

Prompts influence decisions; deterministic gates restrict execution. If policy says identity verification is required before a financial action, the backend should reject the action when verification is absent. Writing “always verify first” does not create a permission boundary. Hooks can intercept calls, but the final service should enforce its own authorization too.

Design financial workflows as **read → propose → authorize → commit → reconcile**. A proposal includes exact account, order, amount, currency, policy version, and expiry. An approval references a hash of that proposal. If the model changes the amount after approval, the hash no longer matches. Persist approval and execution state in a transactional store.

Use idempotency to reduce duplicate side effects during retries. An idempotency key identifies one logical action. Replaying the same key with the same parameters returns the original result; reusing it with different parameters should fail. The example below is an in-memory simulation, so it demonstrates semantics but not durability or atomic concurrency. A real payment service must perform the deduplication and write atomically.

Do not confuse an authenticated login with completed action-specific verification. Whether additional verification is needed depends on policy and risk. Do not trust the model to write `verified=True`; trusted application events establish that fact.

**Human review triggers:** explicit request for a human, policy gap, ambiguous identity, inconsistent authoritative data, irreversible high-impact action outside delegated limits, or repeated inability to progress. Frustration alone need not mean the problem is unresolvable, but the user's explicit human request should be honored.

**Exercise:** A refund request times out. Is retrying safe?

**Answer:** Only after considering whether it already committed. Use an idempotency key and status lookup/reconciliation. A network timeout means the caller does not know the outcome, not that the server did nothing.


````python
@dataclass
class TrustedContext:
    customer_id: str
    verified: bool
    scopes: set

refund_ledger = {}
def commit_refund(ctx, order_customer, amount_minor, currency, key):
    if not ctx.verified or "refund:write" not in ctx.scopes:
        raise PermissionError("Verification and refund scope required")
    if order_customer != ctx.customer_id:
        raise PermissionError("Customer does not own this order")
    if type(amount_minor) is not int or not 0 < amount_minor <= 50000 or currency != "AUD":
        raise ValueError("Outside delegated policy")
    parameters = (ctx.customer_id, amount_minor, currency)
    if key in refund_ledger:
        old = refund_ledger[key]
        if old["parameters"] != parameters:
            raise ValueError("Idempotency key reused for a different action")
        return old
    record = {"parameters": parameters, "status": "committed", "receipt": f"R-{len(refund_ledger)+1}"}
    refund_ledger[key] = record
    return record

trusted = TrustedContext("C-1", True, {"refund:write"})
first = commit_refund(trusted, "C-1", 8500, "AUD", "case-101-refund")
second = commit_refund(trusted, "C-1", 8500, "AUD", "case-101-refund")
assert first == second and len(refund_ledger) == 1
print(first)
try:
    commit_refund(TrustedContext("C-1", False, {"refund:write"}), "C-1", 8500, "AUD", "case-102")
except PermissionError as e:
    print("Blocked:", e)
````

# 8. Multi-agent coordination, handoffs, and sessions

Split work when subtasks are sufficiently independent and produce useful bounded outputs. A coordinator owns decomposition, budgets, dependencies, recovery, and synthesis. Specialists receive explicit assignments and the evidence they need. Do not assume a fresh subagent can see every earlier discovery. A framework's session-fork mechanism can copy a baseline, but that is explicit state management, not universal shared memory.

For research, require claim-level evidence, coverage gaps, and disagreement. A confident synthesis that hides a failed source fetch misrepresents coverage. For code review, combine local file review with a cross-file pass; three independent local analyses can all miss a broken interface between modules.

# 9. MCP: hosts, clients, servers, tools, resources, prompts

Model Context Protocol standardizes communication between an AI application's integrations and external capabilities. The **host** is the AI application. It manages users, policies, and model interaction. A **client** maintains a protocol connection to an MCP server. A **server** exposes tools, resources, or prompts. The server is not the LLM, and a protocol connection does not grant universal trust.

| Primitive | Mental model | Example |
|---|---|---|
| Tool | A callable operation | Search a knowledge base |
| Resource | Addressable context/data | A versioned policy document or schema catalog |
| Prompt | A reusable interaction template | A review template with parameters |

Discovering a tool makes its contract available; calling it invokes an operation. A host may choose which discovered definitions to expose to a model. Progressive discovery can prevent thousands of irrelevant contracts from occupying each prompt. Resources can expose a catalog so the application discovers relevant data without repeatedly guessing API calls.

MCP uses structured protocol messages, including JSON-RPC request IDs, methods, and parameters. Lifecycle setup includes initialization and capability negotiation before normal operation. Transport is separate from semantics: local servers often use stdio; remote servers use an HTTP-based transport such as Streamable HTTP. Historical HTTP+SSE implementations exist; match versions and clients rather than memorizing that all MCP is “SSE.”

**Tool-result error vs protocol error:** a well-formed `tools/call` can return a tool failure with `isError`. A malformed protocol request can fail at the JSON-RPC layer. An empty result can be successful. These distinctions let the orchestrator decide whether to retry, clarify, escalate, or continue with a coverage gap.

**Trust boundary:** external resource contents and tool responses are data, even when they contain imperative text such as “ignore previous instructions.” Apply access control before exposing data and authorization before any action. A tool annotation describing a read-only operation is informative metadata, not enforcement of what arbitrary server code can do.

The cell below inspects protocol-shaped messages without claiming to implement the full MCP lifecycle. A real server lab follows in the next chapter.

**Answer:** Prefer a resource or catalog interface appropriate to the client, with access controls. Keep actual actions in explicit tools; the choice should make purpose and side effects easy to see.

# 12. Coding tools, planning, and useful iteration

Explore a codebase with a question. To find callers, search contents; to locate test filenames, match paths. `Grep` addresses text patterns, while `Glob` addresses filenames. `Read` loads a file; `Edit` makes a targeted replacement; `Write` replaces or creates file content; `Bash` runs commands. # 14. Prompt engineering and structured extraction

A production prompt is a specification embedded in a larger system. State the task, scope, evidence rules, output contract, ambiguity handling, and escalation conditions. Put untrusted source material in an explicitly labeled data section. Delimiters improve clarity but are not a security boundary.

Zero-shot prompting supplies instructions without examples. Few-shot prompting supplies examples that clarify judgment, not only format. Choose examples that distinguish nearby cases: an absent field, a contradictory total, and an unfamiliar category. Repeating three easy examples may teach the format while leaving the real ambiguity unresolved.

For review, define evidence and exclusion criteria. “Report dangerous code” invites subjective output. “Report a path where attacker-controlled input reaches an unrestricted shell command; include the input and command path” is actionable. “Do not flag stylistic preference” prevents a high false-positive category from drowning useful findings.

Structured output has three validation layers: parsing; schema/type constraints; semantic/business constraints. A schema can require `stated_total_minor` and `items`; arithmetic checks whether item amounts sum to the stated total. Evidence checks whether those values actually appear in the source. Missing evidence should yield `null`, `unclear`, or review, not fabricated compliance with a required field.

Current supported Claude APIs expose JSON Schema through `output_config.format`; supported strict tool calls use `strict: true`. Ordinary `tool_use` with a schema is not universally a guarantee of strict conformance. The July guide's tool-extraction pattern remains useful, but distinguish the explicit strict capability from best-effort tool arguments. Always validate business semantics.

The reasoning you need for review is a concise rationale and evidence, not a demand for private hidden chain-of-thought. You can ask the model to analyze carefully and provide a short justification. Intermediate scratch work need not become customer-facing output or durable logs.

**Retry strategy:** provide the source, failed candidate, and precise errors. Bound repair attempts. If the date is absent, retrying cannot reveal it; seek another authorized source or return missingness. If the source itself has a contradictory total, do not “repair” reality by silently changing a number.

Reference: [structured outputs](https://platform.claude.com/docs/en/build-with-claude/structured-outputs), [prompt best practices](https://platform.claude.com/docs/en/build-with-claude/prompt-engineering/overview).


````python
invoice = {"invoice_id": "INV-7", "currency": "AUD", "date": None,
           "items_minor": [1200, 1800], "stated_total_minor": 3200}

def invoice_errors(record):
    errors = []
    if record.get("currency") not in {"AUD", "USD"}:
        errors.append("Unsupported currency")
    amounts = record.get("items_minor")
    if not isinstance(amounts, list) or any(type(x) is not int or x < 0 for x in amounts):
        errors.append("items_minor must contain non-negative integers")
    elif sum(amounts) != record.get("stated_total_minor"):
        errors.append(f"Item sum {sum(amounts)} differs from stated total {record.get('stated_total_minor')}")
    return errors

print(invoice_errors(invoice))
assert invoice_errors(invoice)
fixed_candidate = dict(invoice, stated_total_minor=3000)
assert not invoice_errors(fixed_candidate)
print("Arithmetic passed; source fidelity still needs checking.")

EXTRACTION_SCHEMA = {
    "type": "object",
    "properties": {
        "invoice_id": {"type": ["string", "null"]},
        "currency": {"type": "string", "enum": ["AUD", "USD", "unclear"]},
        "date": {"type": ["string", "null"]},
        "total_minor": {"type": ["integer", "null"]}},
    "required": ["invoice_id", "currency", "date", "total_minor"],
    "additionalProperties": False}

if require_live():
    extraction = anthropic_http("/v1/messages", {
        "model": CLAUDE_MODEL, "max_tokens": 400,
        "system": "Extract only stated facts. Use null for absent fields. Amounts are minor currency units.",
        "messages": [{"role": "user", "content": "Invoice INV-7. AUD 30.00. No date is shown."}],
        "output_config": {"format": {"type": "json_schema", "schema": EXTRACTION_SCHEMA}}})
    if extraction["stop_reason"] != "end_turn":
        raise RuntimeError("Extraction did not complete normally")
    texts = [b["text"] for b in extraction["content"] if b["type"] == "text"]
    print(json.loads("".join(texts)))
else:
    print("Live structured extraction skipped.")
````

# 15. Context budgets, compaction, caching, and memory

Every request competes for a finite context budget. Count system instructions, tool definitions, messages, retrieved evidence, tool results, and relevant overhead. Reserve output capacity and model-specific thinking requirements according to the current model documentation. A hand estimate is useful for planning but does not replace the provider's token-count endpoint.

Let $W$ denote an illustrative total budget. A simplified condition is

$$S+H+D+R+B\leq W,$$

where $S$ is stable instructions, $H$ history, $D$ tool definitions, $R$ retrieved evidence, and $B$ reserved generation space. Exact request validation rules depend on model and API. If $W=32{,}000$, $S=3{,}000$, $H=12{,}000$, $D=2{,}000$, and $B=5{,}000$, at most $10{,}000$ remains for evidence under this simplified allocation.

Do not compact by repeatedly compressing everything into vague prose. Preserve authoritative facts separately: exact IDs, amounts, statuses, dates, open approvals, evidence references, and work already committed. Summaries should identify what is uncertain and which source/version supports a fact. A scratchpad is useful only if it is maintained, access-controlled, and refreshed when its sources change.

Tool outputs often include irrelevant detail. Return a small task-relevant projection instead of the whole customer record. This improves context efficiency and data minimization. Do not omit information required for the decision merely to hit a token target. Large retrieval dumps can bury the decisive clause among unrelated chunks.

Prompt caching reuses eligible prompt prefixes; it does not persist arbitrary business memory or validate facts. Put stable tools/instructions/evidence before changing user input when that fits the request design. Cache hits depend on exact supported prefix matching, sufficient length, configured breakpoints, time window, and platform behavior. Reusing a stale cached policy is still using a stale policy.

**Memory taxonomy:** conversation context helps the current invocation; a summary helps later invocations; a database stores durable records; an approved knowledge source stores policy; a cache reduces repeated work. These solve different problems.

**Exercise:** A conversation remembers “refund roughly eighty dollars.” What should be persisted instead?

**Answer:** Order ID, amount in integer minor units, currency, eligibility evidence, policy version, transaction status, receipt, and provenance. A narrative summary cannot replace a ledger.

References: [context windows](https://platform.claude.com/docs/en/build-with-claude/context-windows), [count tokens](https://platform.claude.com/docs/en/build-with-claude/token-counting), [prompt caching](https://platform.claude.com/docs/en/build-with-claude/prompt-caching).

# 16. Batching, streaming, and extraction at scale

Synchronous calls fit interactive and blocking decisions. Streaming improves time to first visible output, but does not prove the final output will be valid. A UI must handle incomplete content and interruption. Never execute a sensitive tool based on a partially streamed argument before the complete argument has been parsed and authorized.

Message Batches process independent requests asynchronously and currently offer a 50% token-price discount relative to standard usage. They are useful for latency-tolerant jobs, not a guaranteed low-latency pre-merge gate. Read current limits and expiration rules before planning a deadline. A batch request can include a prior conversation, but it does not let your own process execute a newly requested client tool halfway through that individual request; that requires later application orchestration.

Each batch item needs a unique `custom_id`. Correlate by that ID, never assume result order matches input order. Track success, failure, expiration, and cancellation per item. Selectively retry failed items after diagnosing the reason. If one document exceeded context limits, resubmitting it unchanged repeats the problem.

A batch architecture includes input validation, job manifest, submission, status retrieval, result persistence, semantic validation, review queue, and idempotent downstream publishing. An accepted submission is not a completed job. Persist your manifest before submission so recovery can associate results with inputs.

**Deadline math:** if submissions wait at most $a$ hours and processing may take up to $b$ hours under a chosen assumption, the completion envelope is $a+b$ plus result-validation time. This is a planning bound under those assumptions, not a provider availability guarantee. For $a=4$, $b=24$, and two hours of validation, plan for 30 hours and handle expired items explicitly.

**Exercise:** A batch has 900 valid results and 100 errors. Should you rerun all 1,000?

**Answer:** Persist the 900 validated results, diagnose the 100 by failure category, correct inputs/configuration where needed, then retry only relevant failed items with the same logical document identities. Avoid duplicate downstream effects.

Reference: [batch processing](https://platform.claude.com/docs/en/build-with-claude/batch-processing), [streaming](https://platform.claude.com/docs/en/build-with-claude/streaming).

# 17. Reliability, ambiguity, recovery, and Foundations synthesis

Reliability means useful, safe completion under normal load and foreseeable failure. A successful HTTP response is insufficient: measure task outcome, source fidelity, escalation, and side effects. Define which partial outputs are useful and which operations must fail closed.

Classify failure before retrying. A timeout or temporary overload may be retryable with bounded backoff. A schema error needs corrected input. Missing permission needs authorization change or escalation. Policy rejection needs another permitted workflow. Missing source evidence needs retrieval or an honest unknown. Repeating the same request does not solve all five.

Exponential backoff can use

$$d_i=\min(d_{\max},d_0 2^i)+J_i,$$

where $d_i$ is delay before retry $i$ and $J_i$ is randomized jitter. Respect provider retry guidance and overall deadlines. Account for SDK automatic retries so your outer loop does not unintentionally multiply attempts. Only retry side effects with appropriate idempotency/reconciliation.

Recovery should preserve identity and progress. A crash manifest can contain job ID, input versions, completed stages, pending stages, tool outcomes, and verified action receipts. On resume, validate which facts remain current and which need refresh. A saved approval with expired scope cannot be resurrected because the conversation remembers it.

**Foundations task checklist:** implement protocol control; coordinate and delegate; pass context; gate workflows; use hooks; decompose tasks; manage sessions; define tools; classify errors; scope tool access; connect MCP; choose coding tools; scope instructions; package skills; target path rules; plan changes; iterate; automate CI; define review criteria; use examples; structure output; validate extraction; batch jobs; independently review; preserve context; escalate; propagate partial failure; explore codebases; calibrate review; preserve attribution.

This compact checklist maps the guide's 30 task areas to the chapters above. For exact task statements and knowledge/skill lists, use the official PDF rather than assuming this paraphrase is a replacement.

**Six-family drill:** For support, trace ownership and escalation. For coding, inspect scope and planning. For research, trace evidence and gaps. For developer tools, choose narrow capabilities. For CI, enforce non-interactive structured output and useful findings. For extraction, separate schema, arithmetic, and evidence validation.

**Exercise:** A search timeout becomes `[]`, and the final report says “no evidence exists.” Diagnose.

**Answer:** The tool collapsed failed access into a successful empty result. Preserve an error status and coverage gap; the report can say the source could not be checked, not that it contained no evidence.

References: [API errors](https://platform.claude.com/docs/en/api/errors), [rate limits](https://platform.claude.com/docs/en/api/rate-limits).

# 19. Reference patterns, model selection, and decomposition

Choose a pattern by how much uncertainty exists in the next step. A fixed workflow is appropriate when stages and acceptance conditions are predictable. An agent is appropriate when investigation depends on discoveries. RAG supplies evidence; it does not dictate whether orchestration is fixed or agentic. Multi-agent execution supplies specialization; it is not automatically more accurate.

| Pattern | Good fit | Cost or risk |
|---|---|---|
| Single model call | Classification, bounded transformation | Limited verification and context |
| Deterministic workflow | Extraction → validation → review | Less flexible for open investigation |
| RAG-augmented response | Answers over current authorized documents | Retrieval quality and freshness become critical |
| Tool-using agent | Investigation across uncertain paths | More turns, state, and permission risk |
| Coordinator with specialists | Independent evidence domains | Handoff loss, duplicate work, coordination cost |
| Human-approved action | Consequential state changes | Review capacity and approval latency |

Select models using representative measurements. Compare accuracy by task segment, supported features, latency distribution, context capability, cost, and platform availability. Model-family names are useful orientation, but a larger model does not universally improve every pipeline. If retrieval supplies the wrong policy, model size is not the first fix.

A **cascade** uses a cheaper path for suitable requests and escalates uncertain or difficult ones. The routing condition must be evaluated. Self-reported confidence alone is a weak gate. Better signals can include evidence availability, validator outcomes, known task class, or calibrated classifier scores. Confirm that hard cases actually reach the stronger path.

Weighted scoring can expose trade-offs, but only after mandatory constraints. If $s_{ij}$ is candidate $i$'s normalized score on dimension $j$ and $w_j$ its importance, then

$$U_i=\sum_j w_j s_{ij},\qquad \sum_j w_j=1.$$

A candidate that violates a required data boundary is ineligible even if it has the highest utility score. Arbitrary weights can manufacture a desired winner; record the stakeholders' rationale and inspect sensitivity.

**Answer:** Clarify delegated decisions, risk/compliance obligations, evidence, review needs, and acceptable errors. An assistant supporting an existing controlled decision process may be viable; a blanket autonomous commitment requires much stronger justification and controls.

# 20. RAG from ingestion to grounded answers

Retrieval-augmented generation uses retrieved information as evidence for a model response. In a conventional RAG application, the calling program runs retrieval then sends selected evidence to the model. In agentic retrieval, the model requests a search tool and the application executes it. Both are valid; neither means model weights directly query your database without an integration.

The ingestion pipeline loads documents, extracts text or structure, cleans extraction artifacts, preserves provenance, chunks content, computes searchable representations, indexes them, and validates retrieval. Store document ID, version, effective date, section/page, tenant, access metadata, and content hash. Detect deletion and revocation as well as updates.

Chunking is a semantic trade-off. Small chunks are precise but may separate an exception from the rule. Large chunks preserve local meaning but add irrelevant text and consume context. Heading-aware chunks often fit policy documents better than a fixed character split. Tables may require structural extraction so row labels remain associated with values. Overlap helps continuity but can duplicate content and inflate apparent evidence.

Retrieval options include exact metadata filters, lexical search, embeddings, hybrid retrieval, reranking, and graph/structured queries. Exact identifiers and product codes often need lexical or database lookup. Meaning-based queries may benefit from dense vectors. A numeric aggregation should often run against structured data, not ask a model to add values found in prose.

Cosine similarity measures vector direction:

$$\operatorname{cos}(q,d)=\frac{q^\top d}{\|q\|_2\|d\|_2}.$$

For $q=(1,1)$ and $d=(2,0)$, the dot product is 2 and norms are $\sqrt{2}$ and 2, giving $1/\sqrt{2}\approx0.7071$. A similarity of 0.7071 is **not** a 70.71% probability that the document is correct. It is a representation-space score. Zero vectors require special handling.

Embeddings represent data for similarity; they do not enforce access rights. Filter unauthorized candidates before they can reach the model or user. Apply current authorization again at retrieval time, since user membership and document permissions can change after indexing.

Generation should answer from selected evidence, cite source locations, detect conflicts, and abstain when evidence is inadequate. A citation is useful only if it supports the associated claim. Merely attaching a URL to a fabricated sentence is not grounding.

**Exercise:** The assistant becomes wrong immediately after an ingestion change, while model version and latency are stable. What should you inspect first?

**Answer:** Extraction quality, chunk boundaries, index coverage/version, query-document representation compatibility, filters, and retrieved chunks for failing queries. Verify the pipeline before blaming model capacity.

References: [contextual retrieval](https://www.anthropic.com/news/contextual-retrieval), [retrieval cookbook](https://github.com/anthropics/anthropic-cookbook), [embeddings documentation](https://platform.claude.com/docs/en/build-with-claude/embeddings). Choose a separate appropriate embedding provider rather than assuming a generative model endpoint is an embedding endpoint.


````python
def cosine(a, b):
    if len(a) != len(b):
        raise ValueError("Dimensions must match")
    na = math.sqrt(sum(x*x for x in a))
    nb = math.sqrt(sum(x*x for x in b))
    return sum(x*y for x,y in zip(a,b))/(na*nb) if na and nb else 0.0

assert math.isclose(cosine([1,1], [2,0]), 1/math.sqrt(2))

# Runnable bag-of-words retrieval. These are NOT learned embeddings.
documents = [
    {"id":"P1", "tenant":"A", "version":"1", "text":"Refunds require verified order ownership."},
    {"id":"P2", "tenant":"A", "version":"1", "text":"Address changes require account verification."},
    {"id":"P3", "tenant":"B", "version":"1", "text":"Confidential refunds policy for another tenant."}]
def words(text):
    import re
    return re.findall(r"[a-z]+", text.lower())
vocabulary = sorted(set(w for d in documents for w in words(d["text"])))
def bow(text):
    counts = Counter(words(text))
    return [counts[w] for w in vocabulary]
def retrieve(query, tenant, k=2):
    authorized = [d for d in documents if d["tenant"] == tenant]
    ranked_docs = sorted(authorized, key=lambda d: cosine(bow(query),bow(d["text"])), reverse=True)
    return [{**d, "score": cosine(bow(query),bow(d["text"]))} for d in ranked_docs[:k]]

retrieved = retrieve("refunds ownership", "A")
assert retrieved[0]["id"] == "P1"
assert all(d["tenant"] == "A" for d in retrieved)
print(retrieved)
````

# 22. Evaluation: accuracy, retrieval, calibration, and experiments

An evaluation set turns “seems good” into evidence. Define tasks, accepted answers or rubrics, relevant sources, forbidden actions, and failure severity. Include common cases, ambiguity, missing evidence, conflicting documents, unauthorized access, malformed tools, and adversarial source content. Keep development examples separate from a held-out test set.

Measure components as well as end-to-end outcomes. Retrieval can miss evidence; generation can misread correct evidence; tools can fail; permission checks can be wrong; a UI can display an unsupported fragment. Component metrics localize the failure. End-to-end tests show whether the user outcome is acceptable.

For binary findings, let $TP$, $FP$, $FN$, and $TN$ represent true/false positives/negatives:

$$\mathrm{Precision}=\frac{TP}{TP+FP},\quad
\mathrm{Recall}=\frac{TP}{TP+FN},\quad
F_1=\frac{2PR}{P+R}.$$

With $TP=40$, $FP=10$, and $FN=20$, precision is $0.8$, recall is $2/3$, and $F_1\approx0.7273$. A reviewer with many false positives may lose developer trust even if recall is strong. In a safety detector, missed dangerous cases may have much greater cost than unnecessary review. Choose metrics by consequences.

Retrieval recall at $k$ is the fraction of known relevant items found among the top $k$. Reciprocal rank is $1/r$, where $r$ is the first relevant result's rank; mean reciprocal rank averages this over queries. These quantify retrieval ordering, not answer truth.

Segment metrics by document type, field, language, task complexity, and user group where appropriate. Overall 98% extraction accuracy can conceal poor performance on a rare but consequential invoice field. Build human review sampling into “high-confidence” segments too.

**Calibration:** if outputs assigned 0.9 confidence are correct only 0.6 of the time, confidence is overestimated. Measure empirical correctness within score bins. A Brier score for binary outcomes is $\frac1n\sum_i(p_i-y_i)^2$; lower is better. A model's self-description is not calibrated merely because it uses decimal scores.

Mix deterministic validators, expert labels, task-based scoring, and appropriately calibrated model judges. Judges can share model blind spots or prefer verbose answers; blind candidate identity and validate rubric agreement with humans. Repeated sampling helps assess nondeterministic regressions.

**A/B testing:** randomly assign suitable units to variants while keeping other differences controlled. For observed success rates $\hat p_A$ and $\hat p_B$, an approximate independent-sample standard error is

$$SE=\sqrt{\frac{\hat p_A(1-\hat p_A)}{n_A}+\frac{\hat p_B(1-\hat p_B)}{n_B}}.$$

An approximate 95% interval for the difference is $(\hat p_B-\hat p_A)\pm1.96SE$. This formula assumes sufficient independent observations; repeated users, low counts, sequential peeking, and multiple comparisons need a stronger design. Statistical improvement must also be practically useful and preserve safety constraints.

**Exercise:** A new prompt raises accuracy by 2% but doubles p95 latency and increases unauthorized-tool attempts. Is it a clear winner?

**Answer:** No. Compare the full acceptance criteria, uncertainty, and risk. A single quality aggregate cannot override safety or latency gates.

Reference: [evaluation design](https://platform.claude.com/docs/en/test-and-evaluate/develop-tests).


````python
def classification_metrics(tp, fp, fn):
    p = tp/(tp+fp) if tp+fp else 0.0
    r = tp/(tp+fn) if tp+fn else 0.0
    return {"precision": p, "recall": r, "f1": 2*p*r/(p+r) if p+r else 0.0}
print(classification_metrics(40,10,20))

def retrieval_metrics(ranking, relevant, k):
    hits = set(ranking[:k]) & set(relevant)
    ranks = [i+1 for i, doc in enumerate(ranking) if doc in relevant]
    return {"recall_at_k": len(hits)/len(relevant) if relevant else None,
            "reciprocal_rank": 1/min(ranks) if ranks else 0.0}
print(retrieval_metrics(["P2","P1","P7"], {"P1"}, 2))

probabilities, outcomes = [0.9,0.9,0.2,0.1], [1,0,0,0]
brier = sum((p-y)**2 for p,y in zip(probabilities,outcomes))/len(outcomes)
print("Brier score:", brier)

def difference_interval(success_a, n_a, success_b, n_b):
    pa, pb = success_a/n_a, success_b/n_b
    se = math.sqrt(pa*(1-pa)/n_a + pb*(1-pb)/n_b)
    return {"difference": pb-pa, "approx_95_ci": (pb-pa-1.96*se,pb-pa+1.96*se)}
print(difference_interval(850,1000,880,1000))
````

# 23. Observability, latency, throughput, and availability

Observability answers what happened, where time and money were spent, and why a decision failed. Log request/job IDs, component spans, model and prompt versions, input/index versions, stop reason, token usage, tool category, authorization outcome, validation errors, and escalation reason. Redact sensitive payloads; full transcripts should not be the default telemetry format.

Use traces to distinguish slow retrieval from slow model generation and slow downstream tools. Monitor time to first token separately from total response time. Track successful completion, refusal, timeout, cancellation, incomplete output, and semantic validation failure. Latency among successful calls alone hides slow failures.

For a sequential flow, $L_{\text{total}}=L_{\text{auth}}+L_{\text{retrieve}}+L_{\text{model}}+L_{\text{validate}}$. Independent parallel branches contribute roughly their maximum latency plus coordination overhead, not their sum. Dependent stages cannot safely be parallelized simply to improve a chart.

Percentiles describe a distribution. p95 is the value at or below which roughly 95% of observations fall under a chosen calculation. A mean of two seconds can coexist with a p95 of twelve seconds. Tail delays often grow when dependency calls multiply or systems approach saturation.

Little's law gives a steady-state relation

$$N=\lambda L,$$

where $N$ is average work in the system, $\lambda$ arrival rate, and $L$ average time spent. If requests arrive at 10 per second and average completion takes 3 seconds, average in-flight work is 30. This is not a safe maximum-concurrency setting by itself; allow for burstiness, queueing, quotas, and observed tails.

Under an independence assumption, a series of services with availabilities $a_i$ has approximate availability $\prod_i a_i$. Three services at 99.9% each yield about 99.7003% end-to-end availability, not 99.9%. Shared failures violate independence. Retries can improve completion for transient failure while increasing latency and load.

An SLO of 99.9% availability over 30 days allows roughly $30\times24\times60\times0.001=43.2$ minutes of error budget if measured by time. Request-based availability uses eligible request counts instead. Define the indicator explicitly, including exclusions and measurement source.

**Exercise:** The model is fast but customers report slow responses. What next?

**Answer:** Trace the full path: client/network, authentication, queue, retrieval, tool calls, validation, UI rendering, and retries. Inspect p95/p99 rather than the model mean alone.


````python
def nearest_rank(values, percentile):
    if not values or not 0 < percentile <= 100:
        raise ValueError("Need nonempty data and percentile in (0,100]")
    ordered = sorted(values)
    return ordered[math.ceil(percentile/100*len(ordered))-1]

latencies = [1.0 + i*0.05 for i in range(90)] + [10.0]*10
print("Mean:", statistics.mean(latencies), "p95:", nearest_rank(latencies,95))
print("Average in flight:", 10*3)
print("Series availability:", 0.999**3)
print("30-day time error budget in minutes:", 30*24*60*(1-0.999))
trace_event = {"request_id": "req-lab", "stage": "retrieve", "latency_ms": 52,
               "tenant_hash": hashlib.sha256(b"synthetic-A").hexdigest()[:12],
               "index_version": "v1", "authorized_chunks": 2}
print(trace_event)
````

# 24. Token economics, cache break-even, and routing

Prices should be parameters, not hidden constants. Let $I$ and $O$ denote ordinary input/output tokens, and $r_i$, $r_o$ prices per million tokens. Basic request cost is

$$C=\frac{I r_i+O r_o}{10^6}.$$

If a fictional model costs USD 3 per million input tokens and USD 15 per million output tokens, a 4,000-input/800-output call costs $(4{,}000\cdot3+800\cdot15)/10^6=0.024$ dollars. These are teaching prices, not a claim about any current Claude model.

For cache accounting, separate uncached input, cache creation, cache reads, and output. Let $P$ be a repeated prefix, $D$ dynamic input per request, $n$ requests, $r_w$ cache-write price, and $r_c$ cache-read price:

$$C_{\text{cached}}=\frac{P r_w+(n-1)P r_c+nD r_i+nO r_o}{10^6}.$$

The comparable non-cached prefix cost is $nPr_i/10^6$. Caching saves prefix cost when $r_w+(n-1)r_c<nr_i$, or

$$n>\frac{r_w-r_c}{r_i-r_c}$$

when $r_i>r_c$. If an illustrative write costs $1.25r_i$ and read costs $0.1r_i$, the threshold is $1.15/0.9\approx1.278$, so two matching requests can break even under these assumptions. This ignores minimum cacheable size, TTL, misses, and platform details; use observed billing fields and current documentation.

For a model cascade, expected model cost is $c_s+p_e c_l$, where $c_s$ is small-path cost and $p_e$ the escalation fraction to an additional large-path cost $c_l$. If escalation repeats a long prompt, savings may be less than expected. Measure successful-task cost, not cost per attempted invocation.

Total ownership also includes retrieval/embedding compute, storage, tool calls, network, monitoring, human review, engineering, and incidents. A reduction in model price that increases review time can make the workflow more expensive.

Optimization sequence: reduce unnecessary turns; improve retrieval relevance; trim redundant tool fields; bound output; reuse stable prefixes; batch latency-tolerant jobs; test cheaper models; measure the full workflow again. Do not strip essential policy or disable validation to reduce tokens.

**Exercise:** 10,000 requests cost USD 0.02 each, but 20% require USD 1 of review. What dominates?

**Answer:** Model cost is USD 200; review is USD 2,000. Improving first-pass validity could matter more than shaving 10% off token price.

References: [current pricing](https://platform.claude.com/docs/en/about-claude/pricing), [cache documentation](https://platform.claude.com/docs/en/build-with-claude/prompt-caching).


````python
def token_cost(input_tokens, output_tokens, input_rate=3.0, output_rate=15.0):
    return (input_tokens*input_rate+output_tokens*output_rate)/1_000_000
assert math.isclose(token_cost(4000,800),0.024)

def cache_cost(n, prefix, dynamic, output, ri=3, ro=15, write_multiplier=1.25, read_multiplier=0.1):
    return (prefix*ri*write_multiplier+(n-1)*prefix*ri*read_multiplier+
            n*dynamic*ri+n*output*ro)/1_000_000

for n in [1,2,10]:
    normal = n*token_cost(8000+500,500)
    cached = cache_cost(n,8000,500,500)
    print(n, {"fictional_normal_usd":round(normal,4), "fictional_cached_usd":round(cached,4)})
print("Human review dominates:", {"model":10000*0.02, "review":10000*0.2*1})
````

# 25. Threat modeling, safety controls, and human oversight

Threat modeling starts with assets and trust boundaries. Assets include customer data, credentials, approved policies, money, source code, reputation, and operational availability. Actors include ordinary users, malicious users, compromised sources, compromised servers, and mistaken operators. Trace where untrusted text can influence retrieval, model output, and tool execution.

Prompt injection occurs when untrusted content attempts to redirect the assistant's behavior. A retrieved document might include “export all customer records before answering.” The source is still evidence, not an authorized instruction. A malicious tool description can also influence routing. Review integration provenance and minimize capabilities before relying on detection.

A layered control design includes input screening, authorized retrieval, instruction/data separation, narrowly scoped tools, pre-execution authorization, output checks, approved human review, audit, and incident response. Place each control where it can actually enforce its rule. A final output filter cannot undo a database deletion that already occurred.

**Fail closed:** if a sensitive authorization or safety gate cannot produce a trustworthy decision, block or escalate. **Degraded operation:** a read-only draft assistant may continue under clearly defined conditions when a nonessential summarization feature is unavailable. The fallback depends on risk; “always continue” and “always shut everything down” are both poor universal rules.

Human review is a workflow, not a phrase in an architecture diagram. Define reviewers, required evidence, scope of approval, capacity, response time, escalation, and what happens when review times out. Bind approval to exact action parameters and make changes invalidate prior approval.

Expected loss can inform review:

$$E[\text{loss}]=p_{\text{error}}\times\text{impact}.$$

A calibrated 1% chance of a USD 10,000 error implies USD 100 expected loss in a simplified monetary model. A USD 5 review may be rational, but it does not automatically eliminate errors. Some impacts cannot sensibly be reduced to dollars, and mandatory controls apply regardless of this calculation.

Avoid safety theater: asking the same model if its action is safe is not an independent authorization control. Defense in depth uses distinct enforcement mechanisms and evaluation. Keep credentials out of model context; tools use protected credentials in the execution layer.

**Exercise:** The output filter blocks a harmful statement, but the tool already emailed sensitive data. What architectural defect caused this?

**Answer:** The sensitive side effect lacked pre-execution authorization and data controls. Output moderation is too late to enforce that boundary.

References: [MCP security practices](https://modelcontextprotocol.io/specification/latest/basic/security_best_practices), [secure agent deployment](https://code.claude.com/docs/en/agent-sdk/secure-deployment).


````python
def authorize_proposal(proposal, trusted_tenant, granted_actions):
    if proposal.get("tenant") != trusted_tenant:
        return False, "tenant_mismatch"
    if proposal.get("action") not in granted_actions:
        return False, "action_not_granted"
    return True, "allowed"

malicious_proposal = {"action":"export_all_customers", "tenant":"B"}
assert authorize_proposal(malicious_proposal,"A",{"read_policy"})[0] is False
print(authorize_proposal(malicious_proposal,"A",{"read_policy"}))

proposal = {"action":"refund", "order_id":"O-101", "amount_minor":8500, "currency":"AUD"}
def proposal_digest(value):
    return hashlib.sha256(json.dumps(value,sort_keys=True,separators=(",",":")).encode()).hexdigest()
approved_digest = proposal_digest(proposal)
assert approved_digest != proposal_digest(dict(proposal,amount_minor=9000))
print("Changed action invalidates its prior approval binding.")
````

# 26. Governance, data controls, fairness, and evidence

Governance makes ownership and review explicit. Assign who approves the use case, owns data, accepts residual risk, operates the service, responds to incidents, and authorizes changes. A technical control without an owner or evidence cannot be assessed reliably.

Professional objectives mention regulations such as GDPR, HIPAA, and FedRAMP. These apply to different jurisdictions and contexts: they are not interchangeable certifications for an LLM. Learn how to translate an applicable obligation into system controls rather than memorizing that a platform name makes an application compliant. Legal interpretation and applicability need the appropriate specialists.

Use a control matrix:

| Concern | Example control | Owner | Evidence |
|---|---|---|---|
| Unauthorized data access | Tenant/document authorization | Identity/data team | Denial tests and access logs |
| Excess retention | Retention schedule and deletion workflow | Data owner | Deletion records and policy |
| Sensitive prompt payloads | Minimization and approved routing | Application owner | Data-flow review and sampled telemetry |
| Model change regression | Versioned eval release gate | AI engineering | Test report and approval |
| Consequential tool actions | Scope checks and approval binding | Business service owner | Authorization tests and receipts |
| Operational failure | Runbook and incident ownership | Service operations | Drill record and incident timeline |

Data deletion must consider source storage, vectors/indexes, caches, logs, backups, and derived reports. Revoking a user's access at the source while leaving searchable chunks broadly accessible creates a gap. # 27. Stakeholder decisions, architecture records, rollout, and handoff

Professional architecture includes communicating decisions so other people can fund, approve, and operate the system. Frame a trade-off in stakeholder terms: what users gain, what failures remain, what it costs, who must review, and how it can be reversed. “We use a coordinator with vector retrieval” is an implementation description, not a business decision.

An architecture decision record should contain context, requirements, alternatives, decision, rationale, consequences, and revisit triggers. Record why a workflow was selected over an agent, why one deployment platform fits the approved boundary, and which measured results support model choice. Keep assumptions visible, especially traffic forecasts and review staffing.

Lifecycle stages include discovery, prototype, acceptance evaluation, security/integration review, limited pilot, controlled rollout, operation, and improvement. Define exit criteria for each stage. Prototype success on ten curated cases cannot establish readiness for production permissions.

A rollout can begin in shadow mode: produce recommendations without acting, compare against actual outcomes, and identify gaps. Then enable assisted use with review, followed by selective automation for proven low-risk categories. Use feature flags and a rollback plan. A model rollback may need matching prompt, tool, and index versions to restore the earlier behavior.

Handoff must survive your absence. Include system/data-flow diagrams, run commands, deployment process, identities, quotas, dashboards, incident procedures, approval workflows, known limitations, test corpus ownership, cost limits, and change control. Run a handoff drill: someone else diagnoses a synthetic failure using only the documentation.

**Acceptance is negotiated explicitly:** business owners define useful outcomes; data owners approve sources; security approves boundaries; operations accepts runbooks; engineering maintains tests and deployment. Map disagreement to the relevant decision owner rather than assuming one architect can sign everything off.

**Exercise:** Explain a retrieval reranker to an executive in one sentence.

**Answer:** “It selects more relevant evidence before the answer is written, which can improve answer quality, but adds processing time and cost; our pilot will test whether the improvement meets your response-time target.”

**Reusable ADR outline:** Problem; constraints; options; measurements; decision; risks and controls; owner; rollback; revisit conditions. For example, revisit if p95 latency breaches the budget, the provider route changes, or a newly significant task segment fails evaluation.

## Runnable miniature: retrieval → answer → validation → review decision

The pipeline below uses a deterministic answerer so all parts run offline. It proves the harness behavior under controlled examples; it does **not** validate real LLM quality. Afterward, substitute a live model and build a larger labeled set. The fixture's allowed answer vocabulary is intentionally narrow.

**Exercises:** add a conflicting second policy; require version freshness; introduce a sensitive source; simulate no retrieval; test citation validity; implement a review queue; compare two retrieval configurations. Preserve the distinction between “the fake fixture passed” and “the production model was evaluated.”


````python
def support_pipeline(query, tenant="A"):
    evidence = retrieve(query, tenant, k=1)
    if not evidence or evidence[0]["score"] <= 0:
        return {"status":"review", "answer":None, "citations":[], "reason":"missing_evidence"}
    source = evidence[0]
    answer = source["text"]  # Deterministic extractive baseline, not model generation.
    citations = [{"id":source["id"],"version":source["version"]}]
    allowed_ids = {d["id"] for d in evidence}
    citation_ok = all(c["id"] in allowed_ids for c in citations)
    if not citation_ok:
        return {"status":"review","answer":None,"citations":[],"reason":"invalid_citation"}
    return {"status":"answered","answer":answer,"citations":citations,"reason":None}

pipeline_cases = [
    {"query":"refunds ownership","tenant":"A","expected_status":"answered","source":"P1"},
    {"query":"address changes","tenant":"A","expected_status":"answered","source":"P2"},
    {"query":"astronomy","tenant":"A","expected_status":"review","source":None},
    {"query":"refunds","tenant":"Z","expected_status":"review","source":None}]
for case in pipeline_cases:
    got = support_pipeline(case["query"],case["tenant"])
    assert got["status"] == case["expected_status"]
    if case["source"]:
        assert got["citations"][0]["id"] == case["source"]
    print(case["query"], got)
print("Synthetic harness checks passed. Real-model quality remains unevaluated.")
````

````python
def grade_practice(answers, exam="F"):
    bank = F_BANK if exam == "F" else P_BANK if exam == "P" else None
    if bank is None:
        raise ValueError("exam must be F or P")
    totals, correct, missed = Counter(), Counter(), []
    for i,item in enumerate(bank,1):
        qid=f"{exam}{i:02d}"
        domain=item[0]
        expected=PRACTICE_ANSWER_KEYS[exam][qid]
        totals[domain]+=1
        if answers.get(qid, "").upper()==expected:
            correct[domain]+=1
        else:
            missed.append(qid)
    return {"raw_accuracy":sum(correct.values())/len(bank),
            "by_domain":{d:correct[d]/totals[d] for d in totals},"missed":missed}

# Example after studying: grade_practice({"F01":"A","F02":"B"}, "F")
print("Scoring helper defined. Fill your own answer dictionary after a closed-book attempt.")
````

# 33. Worked engineering problems, pitfalls, and final revision

## Problem 1: cascade quality and cost

Suppose 80% of requests use an easy route with 95% success, and the other 20% use a difficult route with 90% success. If classification into routes is correct, aggregate success is

$$0.8(0.95)+0.2(0.90)=0.94.$$

This does not prove the router is correct. If hard tasks leak into the easy route, actual success changes. Evaluate the routing confusion matrix and conditional route outcomes.

If every request first uses a USD 0.005 classifier/cheap path and 20% additionally use a USD 0.05 route, expected model spend is $0.005+0.2(0.05)=0.015$ dollars per request. Add retrieval, review, and other operations. At 10,000 requests this is USD 150 under the fictional assumptions.

## Problem 2: retries amplify load

If independent failure probability is $p$ and you allow at most $m$ total attempts, completion probability is $1-p^m$. Expected attempts are

$$E[A]=1+p+p^2+\cdots+p^{m-1}=\frac{1-p^m}{1-p}.$$

For $p=0.2$ and $m=3$, completion is $0.992$ and expected attempts $1.24$. This independence assumption is weak during a shared outage; retries can overload dependencies rather than rescue the task. Rate limits, jitter, deadlines, and circuit breakers matter. Authorization and validation errors should not be modeled as random transient failures.

## Problem 3: review staffing

Suppose 2,000 items arrive daily, 15% require review, and each takes four minutes. That is 300 reviews or 1,200 minutes = 20 reviewer-hours daily. One reviewer with six productive review hours cannot absorb it. Staff capacity, reduce unnecessary exceptions without sacrificing quality, or constrain volume. A review queue without capacity planning becomes a bottleneck and a temptation to bypass controls.

## Problem 4: sample uncertainty

An observed accuracy $\hat p=0.95$ from $n=100$ independent cases has approximate standard error $\sqrt{0.95(0.05)/100}\approx0.0218$. A rough normal interval is $0.95\pm0.0427$. Near boundaries or for small samples, methods such as Wilson intervals are better. More importantly, a hundred easy cases do not establish quality on a missing high-risk segment. Sampling representativeness and uncertainty both matter.

## Problem 5: hybrid ranking

Retrievers have incomparable score scales. Reciprocal-rank fusion combines ranks without treating cosine and lexical scores as equal probabilities:

$$\operatorname{RRF}(d)=\sum_j\frac{1}{c+r_j(d)},$$

where $r_j(d)$ is rank in retriever $j$ and $c>0$ dampens high-rank dominance. If a document is absent from a retriever list, that retriever contributes zero. This is optional architecture enrichment, not an assertion that the exams require the formula. Evaluate it against query segments and latency before adopting it.

````python
def retry_math(p, attempts):
    if not 0 <= p < 1 or attempts < 1:
        raise ValueError("Invalid inputs")
    return {"completion_under_independence":1-p**attempts,
            "expected_attempts":sum(p**i for i in range(attempts))}
print(retry_math(0.2,3))
print("Review hours/day:", 2000*0.15*4/60)

def wilson_interval(successes,n,z=1.96):
    if n <= 0 or not 0 <= successes <= n:
        raise ValueError("Invalid counts")
    p=successes/n
    denom=1+z*z/n
    center=(p+z*z/(2*n))/denom
    half=z*math.sqrt(p*(1-p)/n+z*z/(4*n*n))/denom
    return (center-half, center+half)
print("Wilson interval for 95/100:", wilson_interval(95,100))

def reciprocal_rank_fusion(rankings,c=60):
    scores=Counter()
    for ranking in rankings:
        if len(ranking)!=len(set(ranking)):
            raise ValueError("Duplicate document in ranking")
        for rank,doc in enumerate(ranking,1):
            scores[doc]+=1/(c+rank)
    return sorted(scores.items(),key=lambda row:row[1],reverse=True)
print("Fused ranking:",reciprocal_rank_fusion([["P1","P2","P3"],["P2","P1","P4"]]))
````
