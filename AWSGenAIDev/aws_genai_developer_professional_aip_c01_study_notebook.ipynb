## 5. RAG Architecture Cheat Sheet

### Standard production RAG flow

1. Source documents land in S3.
2. Ingestion pipeline validates and preprocesses content.
3. Documents are chunked.
4. Embeddings are generated.
5. Embeddings and metadata are stored in vector store.
6. User query is embedded.
7. Relevant chunks are retrieved.
8. Retrieved context is injected into prompt.
9. FM generates answer.
10. Guardrails and output validation run.
11. Response includes citations/source attribution.
12. Logs, metrics, trace, feedback are stored.

### RAG design decisions

| Decision | What to consider |
|---|---|
| Chunk size | Too small loses context; too large harms retrieval and token cost |
| Overlap | Helps preserve context across chunk boundaries |
| Metadata | Security filtering, document type, source, timestamp, sensitivity |
| Embedding model | Domain/language support, cost, latency, dimension |
| Retrieval | Top-k, similarity threshold, hybrid search, reranking |
| Access control | Filter retrieval by user entitlement |
| Freshness | Scheduled ingestion, event-driven ingestion, sync policies |
| Evaluation | Retrieval precision/recall, answer faithfulness, citation accuracy |

- Managed RAG

Use OpenSearch when the question wants:

```python
# Example: Lambda-style validation for document ingestion

import json
import re

ALLOWED_EXTENSIONS = {".pdf", ".txt", ".md", ".csv", ".json", ".docx"}
MAX_SIZE_MB = 50

def validate_s3_object(bucket, key, size_bytes):
    errors = []
    extension = "." + key.split(".")[-1].lower() if "." in key else ""

    if extension not in ALLOWED_EXTENSIONS:
        errors.append(f"Unsupported file extension: {extension}")

    if size_bytes > MAX_SIZE_MB * 1024 * 1024:
        errors.append(f"File exceeds max size of {MAX_SIZE_MB} MB")

    if re.search(r"/temp/|/draft/", key.lower()):
        errors.append("Draft/temp documents should not be ingested")

    return {
        "is_valid": len(errors) == 0,
        "errors": errors,
        "metadata": {
            "extension": extension,
            "size_mb": round(size_bytes / 1024 / 1024, 2)
        }
    }

# Example event fragment
print(validate_s3_object("my-doc-bucket", "policies/hr-policy.pdf", 10_000_000))
```

## 8. Domain 2 Deep Dive: Implementation and Integration

This domain is about building actual applications.

### 8.1 Agentic AI

Agents are useful when the model needs to:

- Decide which tool/API to call
- Execute multi-step workflows
- Use memory/state
- Interact with enterprise systems
- Complete tasks, not just answer questions

### Agent architecture

1. User request enters API
2. Input guardrail validates request
3. Agent receives instruction + tool schema
4. Agent selects tool
5. Tool executes through Lambda/ECS/API
6. Tool output returns to agent
7. Agent reasons next step
8. Output guardrail validates final response
9. Trace/log is stored

### Agent safety controls

- Limit allowed tools
- Least-privilege IAM role
- Parameter validation in Lambda
- Timeouts and max steps
- Human approval for irreversible actions
- Guardrails on input/output
- Tool result validation
- Audit logging
- Circuit breakers and fallback

```python
# Example tool schema idea for an agent.
# The exact format differs by implementation, but the exam concept is what matters.

get_order_status_tool = {
    "name": "get_order_status",
    "description": "Retrieve order status for a customer order ID.",
    "input_schema": {
        "type": "object",
        "properties": {
            "order_id": {
                "type": "string",
                "description": "The order ID. Must match pattern ORD- followed by digits."
            }
        },
        "required": ["order_id"]
    }
}

def lambda_handler(event, context):
    # Validate parameters strictly. Never trust model-generated tool args blindly.
    order_id = event.get("order_id", "")
    if not order_id.startswith("ORD-"):
        return {"error": "Invalid order_id format"}

    # Call backend safely here.
    return {
        "order_id": order_id,
        "status": "In transit",
        "estimated_delivery": "2026-07-02"
    }
```

```python
# Example serverless API handler for Bedrock Converse API

import json
import boto3

bedrock_runtime = boto3.client("bedrock-runtime")

def handler(event, context):
    try:
        body = json.loads(event.get("body", "{}"))
        user_input = body.get("message", "")

        if not user_input.strip():
            return {
                "statusCode": 400,
                "body": json.dumps({"error": "message is required"})
            }

        response = bedrock_runtime.converse(
            modelId="anthropic.claude-3-5-haiku-20241022-v1:0",
            messages=[
                {
                    "role": "user",
                    "content": [{"text": user_input}]
                }
            ],
            inferenceConfig={
                "temperature": 0.2,
                "maxTokens": 500
            }
        )

        return {
            "statusCode": 200,
            "headers": {"Content-Type": "application/json"},
            "body": json.dumps(response, default=str)
        }

    except Exception as e:
        # In production, log structured details to CloudWatch,
        # but do not leak internal error details to user.
        return {
            "statusCode": 500,
            "body": json.dumps({"error": "Internal server error"})
        }
```

```python
# Example PII redaction concept with Amazon Comprehend.
# Requires AWS permissions and region support.

import boto3

comprehend = boto3.client("comprehend", region_name="us-east-1")

def detect_pii(text: str):
    response = comprehend.detect_pii_entities(
        Text=text,
        LanguageCode="en"
    )
    return response["Entities"]

def simple_redact(text: str, entities):
    # Redact from end to start so offsets do not shift.
    result = text
    for ent in sorted(entities, key=lambda e: e["BeginOffset"], reverse=True):
        begin, end = ent["BeginOffset"], ent["EndOffset"]
        result = result[:begin] + f"[REDACTED_{ent['Type']}]" + result[end:]
    return result
```

```python
# Simple token/cost estimation model for studying.
# Replace prices with current model pricing when planning real workloads.

def estimate_monthly_cost(
    requests_per_day,
    avg_input_tokens,
    avg_output_tokens,
    input_price_per_1k,
    output_price_per_1k,
    days=30
):
    monthly_requests = requests_per_day * days
    input_cost = monthly_requests * avg_input_tokens / 1000 * input_price_per_1k
    output_cost = monthly_requests * avg_output_tokens / 1000 * output_price_per_1k
    return {
        "monthly_requests": monthly_requests,
        "input_cost": round(input_cost, 2),
        "output_cost": round(output_cost, 2),
        "total_cost": round(input_cost + output_cost, 2)
    }

estimate_monthly_cost(
    requests_per_day=10000,
    avg_input_tokens=1200,
    avg_output_tokens=400,
    input_price_per_1k=0.00025,
    output_price_per_1k=0.00125
)
```

| Latency | Does response meet SLO? |
| Cost | Is token usage acceptable? |

### RAG evaluation

| Metric | What it checks |
|---|---|
| Retrieval precision | Are retrieved chunks relevant? |
| Retrieval recall | Did we retrieve all necessary facts? |
| Answer faithfulness | Is answer grounded in retrieved context? |
| Citation accuracy | Do citations support the claim? |
| Context relevance | Is retrieved context useful? |
| Hallucination rate | Unsupported claims |
| Latency | Retrieval + generation time |

|---|---|

```python
# Example schema validation for structured model output

import json
from jsonschema import validate, ValidationError

answer_schema = {
    "type": "object",
    "properties": {
        "answer": {"type": "string"},
        "confidence": {"type": "string", "enum": ["high", "medium", "low"]},
        "sources": {"type": "array", "items": {"type": "string"}},
        "requires_human_review": {"type": "boolean"}
    },
    "required": ["answer", "confidence", "sources", "requires_human_review"]
}

def validate_model_json(raw_text: str):
    try:
        obj = json.loads(raw_text)
        validate(instance=obj, schema=answer_schema)
        return True, obj
    except (json.JSONDecodeError, ValidationError) as e:
        return False, str(e)

sample = '''
{
  "answer": "The policy allows reimbursement with manager approval.",
  "confidence": "medium",
  "sources": ["travel-policy.pdf"],
  "requires_human_review": false
}
'''

validate_model_json(sample)
```

## 16. Troubleshooting Matrix

| Symptom | Likely cause | Best fix |
|---|---|---|
| Hallucinated answer | No grounding or weak retrieval | RAG, citations, confidence, answer-only-from-context prompt |
| Correct docs not retrieved | Bad chunking/embedding/metadata | Re-chunk, tune top-k, improve metadata, hybrid search |
| Too many irrelevant chunks | Top-k too high or poor index | Tune retrieval, reranking, filters |
| Output not JSON | Weak prompt or no validation | Schema prompt + validate + retry/repair |
| High latency | Large model, too much context, slow vector DB | Smaller model, streaming, reduce chunks, optimize index |
| High cost | Large input/output tokens | compression, caching, model routing |
| Prompt works in dev but fails in prod | prompt drift/version mismatch | Prompt Management, regression tests |
| Agent loops forever | no max step/timeout | step limit, stopping condition, Step Functions |
| Agent calls wrong tool | ambiguous tool definitions | better tool descriptions, validation, tool constraints |
| Data leak across users | missing authorization filter | tenant filters in retrieval, IAM, ABAC |
| Throttling | request bursts or quota | exponential backoff, queueing, provisioned throughput |
| Unsafe output | no output guardrail | Bedrock Guardrails + post-processing |
| Sensitive data in logs | logging raw prompts | redaction, log policy, KMS, restricted access |
